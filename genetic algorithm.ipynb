```python
from numpy.random import randint,rand
```

```python
def objective(x):
    return x[0]**2+x[1]**2
```

```python
def decode(bounds,n_bits,bitstr):
    decoded=[] #list()
    largest=2**n_bits
    for i in range(len(bounds)):
         s,e=i*n_bits,(i*n_bits)+n_bits
         substring=bitstr[s:e]
         chars="".join([str(i) for i in substring])
         integer=int(chars,2)

         val=bounds[i][0]+(integer/largest)*(bounds[i][1]-bounds[i][0])
         decoded.append(val)

    return decoded
    
```

```python
def selection(pop,scores,k=3):
    selection=randint(len(pop))

    for i in randint(0,len(pop),k-1):
        if scores[i]<scores[selection]:
            selection=i

    return pop[selection]
```

```python
def crossover(p1,p2,r_cross):
    c1=p1.copy()
    c2=p2.copy()

    if rand()<r_cross:
        pt=randint(1,len(p1)-2)
        c1=p1[:pt]+p2[pt:]
        c2=p2[:pt]+p1[pt:]

    return [c1,c2]
```

```python
def mutation(bitstr,r_mut):

    for i in range(len(bitstr)):
        if rand()<r_mut:
            bitstr[i]=1-bitstr[i]
```

```python
def main(objective,bounds,n_bits,n_iter,n_pop,r_cross,r_mut):

    pop=[randint(0,2,n_bits*len(bounds)).tolist() for i in range(n_pop)]

    best,best_eval=0,objective(decode(bounds,n_bits,pop[0]))

    for i in range(100):
        decoded=[decode(bounds,n_bits,p) for p in pop]
        scores=[objective(d) for d in decoded]

        for i in range(100):
            if scores[i]<best_eval:

                best,best_eval=pop[i],scores[i]

                print(f"> iter {i},new best {decoded[i]} = {scores[i]}")

        selected=[selection(pop,scores) for i in range(n_pop)]

        children=[]

        for i in range(0,n_pop,2):
            p1,p2=selected[i],selected[i+1]
            for i in crossover(p1,p2,r_cross):
                mutation(i,r_mut)
                children.append(i)

        pop=children

        return [best,best_eval]
```

```python
bounds=[[-5.0,5.0],[-5.0,5.0]]
n_bits=16
n_pop=100
r_cross=0.9
r_mut=1.0/(float(n_bits)*len(bounds))
```

```python
print("started.........")
best,score=main(objective,bounds,n_bits,n_iter,n_pop,r_cross,r_mut)
decode=decode(bounds,n_bits,best)
print("completedd")
```

```python
print(f"best sol:{decode}")
print(f"fitness score of best sol:{score:.5f}")
```
