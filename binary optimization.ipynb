```python
from numpy.random import randint,rand
```

```python
def onemax(x):
    return -sum(x)
```

```python
def selection(pop,scores,k=3):
    selection_ix=randint(len(pop))

    for i in randint(0,len(pop),k-1):
        if scores[i]<scores[selection_ix]:
            slection_ix=i

    return pop[selection_ix]
```

```python
def crossover(p1,p2,r_cross):
    c1=p1.copy()
    c2=p2.copy()

    if rand()<r_cross:
        pt=randint(1,len(p1)-2)

        c1=p1[:pt]+p2[pt:]
        c2=p2[:pt]+p1[pt:]

    return [c1,c2]
```

```python
def mutation(bitstring,r_mut):
    for i in range(len(bitstring)):
        if rand()<r_mut:
            bitstring[i]=1-bitstring[i]
```

```python
def genetic_algorithm(ob,n_bits,n_iter,n_pop,r_cross,r_mut):

    pop = [randint(0,2,n_bits).tolist() for i in range(n_pop)]
    best,best_eval=0,ob(pop[0])

    for i in range(n_iter):

        scores=[ob(i) for i in pop]

        for i in range(n_pop):
            
            if scores[i]<best_eval:
                best,best_eval=pop[i],scores[i]
                print("iter %d,new best f(%s)=%.3f"%(i,pop[i],scores[i]))


        selected=[selection(pop,scores) for i in range(n_pop)]
        children=list()

        for i in range(0,n_pop,2):
            p1,p2=selected[i],selected[i+1]

            for i in crossover(p1,p2,r_cross):
                mutation(i,r_mut)
                children.append(i)

        pop=children
        return [best,best_eval]
        
```

```python
n_iter=100
n_bits=20
n_pop=100
r_cross=0.9
r_mut=1.0/float(n_bits)

print("starting....")
best,score=genetic_algorithm(onemax,n_bits,n_iter,n_pop,r_cross,r_mut)
print("completed")
```
