```python
from numpy import asarray,exp
from numpy.random import randn,rand,seed
```

```python
def objective(x):
    return x[0]**2.0
```

```python
def simulated_annealing(objective,bounds,n_iterations,step_size,temp):
    best=bounds[:,0]+rand(len(bounds))*(bounds[:,1]-bounds[:,0])
    best_eval=objective(best)
    curr,curr_eval=best,best_eval
    
    for i in range(n_iterations):
         candidate=curr+rand(len(bounds))*step_size
         cand_eval=objective(candidate)

         if cand_eval < best_eval:
             best,best_eval=candidate,cand_eval
             print(">iteration %d:f(%s)=%05f"%(i,best,best_eval))

         diff=cand_eval - curr_eval
         t=temp/float(i+1)
         metropolis=exp(-diff/t)

         if diff < 0 or rand()<metropolis:
             curr,curr_eval=candidate,cand_eval

    return [best,best_eval]
```

```python
seed(1)
bounds=asarray([[-5.0,5.0]])
n_iterations=1000
step_size=0.1
temp=10

print("startedd")
best,score=simulated_annealing(objective,bounds,n_iterations,step_size,temp)
print("completed")

print(f"best sol:{best}")
print(f"fitness score of best sol:{score:.5f}")
```
